# Data Distribution using different distribution plots

# 1.) Using the Kernel Distribution Plots

### Data Distribution: Kernel Density Estimation (KDE) Plots

In this section, we will use Kernel Density Estimation (KDE) plots to visualize the distributions of numerical features in the Chronic Kidney Disease (CKD) dataset and compare patients with and without CKD.

**Steps:**

1. **Import the necessary libraries:** Use `os` to manage folders, `seaborn` to generate KDE plots, and `matplotlib` to display and save them.

2. **Prepare the data:** Create a copy of `dfnum3`, which contains the numerical features after missing-value imputation. Add the `classification` column from the original dataset (`df`) to distinguish CKD and Not CKD patients.

3. **Prepare the classification labels:** Convert the classification values into readable labels: `CKD` and `Not CKD`.

4. **Create an output folder:** Create a folder named `Kernel_Plots` in the current working directory to store the generated plots.

5. **Select numerical features:** Use the numerical columns from `dfnum3` for visualization.

6. **Generate KDE plots:** Use `sns.kdeplot()` to visualize the distribution of each numerical feature, with separate curves for CKD and Not CKD patients.

7. **Save and display the plots:** Save each plot as a PNG image in the `Kernel_Plots` folder and display it in Jupyter Notebook.

**Interpretation:** Compare the peaks, spread and overlap of the two distributions. These plots help us understand how numerical features differ between CKD and non-CKD patients and identify potentially informative features for classification.

```python

import os
import seaborn as sns
import matplotlib.pyplot as plt

# Use the numerical data after imputation
plot_df = dfnum3.copy()

# Add the CKD classification labels
plot_df['classification'] = df['classification'].replace({
    0: 'Not CKD',
    1: 'CKD',
    'ckd': 'CKD',
    'notckd': 'Not CKD'
}).to_numpy()

# Directory where plots will be saved
# output_dir = '/Users/ahmed/Desktop/CDUT-OBU Classes/CHC6089 Machine Learning-Ahmed/Week 2/kernel_plots '
output_dir = 'kernel_plots'
os.makedirs(output_dir, exist_ok = True)

# Select numerical features
features_to_plot = dfnum3.columns

# Set plot style
sns.set(font_scale = 1.2)
sns.set_style("white")

# Generate and save KDE plots
for feature in features_to_plot:
    plt.figure(figsize = (8, 6))

    sns.kdeplot(
        data = plot_df,
        x = feature,
        hue = 'classification',
        fill = True,
        common_norm = False,
        warn_singular = False
    )

    plt.title(f'Kernel Distribution of {feature.upper()}')
    plt.xlabel(feature.title())
    plt.ylabel('Density')

    plt.tight_layout()

    # Save the plot
    plot_filename = f'kernel_plot_{feature}.png'
    plot_path = os.path.join(output_dir, plot_filename)
    plt.savefig(plot_path, dpi = 300)

    plt.show()
```

# Data split 

```python

from sklearn.model_selection import train_test_split

# Convert classification labels to 0 and 1
y = y.replace({'ckd': 1, 'not ckd': 0, 'notckd': 0}).astype('int64')

# Split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.2,
    random_state = 42,
    stratify = y
)

# Display the size of each dataset
print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)

# Display class distribution
print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

# Feature Scaling (StandardScaler)

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit and transform the training data
x_train_scaled = scaler.fit_transform(x_train)

# Transform the testing data using the same scaler
x_test_scaled = scaler.transform(x_test)

# Display the size of scaled datasets
print("Scaled training data:", x_train_scaled.shape)
print("Scaled testing data:", x_test_scaled.shape)
```

# Data Balancing (SVMSMOTE)

```python
from imblearn.over_sampling import SVMSMOTE

# Create the SVMSMOTE object
smote = SVMSMOTE(random_state=42)

# Apply SVMSMOTE to training data only
x_train_resampled, y_train_resampled = smote.fit_resample(
    x_train_scaled, y_train
)

# Display class distribution before SMOTE
print("Before SMOTE:")
print(y_train.value_counts())

# Display class distribution after SMOTE
print("\nAfter SMOTE:")
print(y_train_resampled.value_counts())

# Display the size of balanced training data
print("\nBalanced training data:", x_train_resampled.shape)
```

# Logistic Regreasion

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
lr_model = LogisticRegression(max_iter = 1000, random_state = 42)

# Train the model
lr_model.fit(x_train_resampled, y_train_resampled)

# Predict the testing data
lr_pred = lr_model.predict(x_test_scaled)

print("Logistic Regression predictions:")
print(lr_pred)
```

# Logistic Regression Evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Calculate evaluation metrics
print("Logistic Regression Evaluation")
print("------------------------------")

print("Accuracy:", accuracy_score(y_test, lr_pred))
print("Precision:", precision_score(y_test, lr_pred, pos_label = 1))
print("Recall:", recall_score(y_test, lr_pred, pos_label = 1))
print("F1-Score:", f1_score(y_test, lr_pred, pos_label = 1))

# Display classification report
print("\nClassification Report:")
print(classification_report(y_test, lr_pred))
```

# Logistic Regression Confusion Matrix

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Calculate confusion matrix
lr_cm = confusion_matrix(y_test, lr_pred, labels=[0, 1])

# Plot confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    lr_cm,
    annot = True,
    fmt = 'd',
    cmap = 'Blues',
    xticklabels = ['Not CKD', 'CKD'],
    yticklabels = ['Not CKD', 'CKD']
)

plt.title('Logistic Regression - Confusion Matrix')
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')

plt.tight_layout()
plt.show()
```

# K-Nearest Neighbors (KNN)

```python
from sklearn.neighbors import KNeighborsClassifier

# Create the KNN model
knn_model = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_model.fit(x_train_resampled, y_train_resampled)

# Predict the testing data
knn_pred = knn_model.predict(x_test_scaled)

print("KNN predictions:")
print(knn_pred)
```

# KNN Evaluation

```python
# Calculate evaluation metrics
print("KNN Evaluation")
print("--------------")

print("Accuracy:", accuracy_score(y_test, knn_pred))
print("Precision:", precision_score(y_test, knn_pred, pos_label=1))
print("Recall:", recall_score(y_test, knn_pred, pos_label=1))
print("F1-Score:", f1_score(y_test, knn_pred, pos_label=1))

# Display classification report
print("\nClassification Report:")
print(classification_report(y_test, knn_pred))
```
