# Data Distribution using different distribution plots

# 1.) Using the Kernel Distribution Plots

### Data Distribution: Kernel Density Estimation (KDE) Plots

In this section, we will use Kernel Density Estimation (KDE) plots to visualize the distributions of numerical features in the Chronic Kidney Disease (CKD) dataset and compare patients with and without CKD.

**Steps:**

1. **Import the necessary libraries:** Use `os` to manage folders, `seaborn` to generate KDE plots, and `matplotlib` to display and save them.

2. **Prepare the data:** Create a copy of `dfnum3`, which contains the numerical features after missing-value imputation. Add the `classification` column from the original dataset (`df`) to distinguish CKD and Not CKD patients.

3. **Prepare the classification labels:** Convert the classification values into readable labels: `CKD` and `Not CKD`.

4. **Create an output folder:** Create a folder named `Kernel_Plots` in the current working directory to store the generated plots.

5. **Select numerical features:** Use the numerical columns from `dfnum3` for visualization.

6. **Generate KDE plots:** Use `sns.kdeplot()` to visualize the distribution of each numerical feature, with separate curves for CKD and Not CKD patients.

7. **Save and display the plots:** Save each plot as a PNG image in the `Kernel_Plots` folder and display it in Jupyter Notebook.

**Interpretation:** Compare the peaks, spread and overlap of the two distributions. These plots help us understand how numerical features differ between CKD and non-CKD patients and identify potentially informative features for classification.

```python

import os
import seaborn as sns
import matplotlib.pyplot as plt

# Use the numerical data after imputation
plot_df = dfnum3.copy()

# Add the CKD classification labels
plot_df['classification'] = df['classification'].replace({
    0: 'Not CKD',
    1: 'CKD',
    'ckd': 'CKD',
    'notckd': 'Not CKD'
}).to_numpy()

# Directory where plots will be saved
# output_dir = '/Users/ahmed/Desktop/CDUT-OBU Classes/CHC6089 Machine Learning-Ahmed/Week 2/kernel_plots '
output_dir = 'kernel_plots'
os.makedirs(output_dir, exist_ok = True)

# Select numerical features
features_to_plot = dfnum3.columns

# Set plot style
sns.set(font_scale = 1.2)
sns.set_style("white")

# Generate and save KDE plots
for feature in features_to_plot:
    plt.figure(figsize = (8, 6))

    sns.kdeplot(
        data = plot_df,
        x = feature,
        hue = 'classification',
        fill = True,
        common_norm = False,
        warn_singular = False
    )

    plt.title(f'Kernel Distribution of {feature.upper()}')
    plt.xlabel(feature.title())
    plt.ylabel('Density')

    plt.tight_layout()

    # Save the plot
    plot_filename = f'kernel_plot_{feature}.png'
    plot_path = os.path.join(output_dir, plot_filename)
    plt.savefig(plot_path, dpi = 300)

    plt.show()
```

# 2.) Using the Violin Plots

```python

import os
import seaborn as sns
import matplotlib.pyplot as plt

# Use the numerical data after imputation
plot_df = dfnum3.copy()

# Add the CKD classification labels
plot_df['classification'] = df['classification'].replace({
    0: 'Not CKD',
    1: 'CKD',
    'ckd': 'CKD',
    'notckd': 'Not CKD'
}).to_numpy()

# Directory where plots will be saved
# output_dir = '/Users/ahmed/Desktop/CDUT-OBU Classes/CHC6089 Machine Learning-Ahmed/Week 2/Violin_Plots'
output_dir = 'violin_plots'

# Create the directory if it doesn't exist
os.makedirs(output_dir, exist_ok = True)

# Select numerical features
features_to_plot = dfnum3.columns

# Set plot style
sns.set(font_scale = 1.2)
sns.set_style("white")

# Generate individual violin plots
for feature in features_to_plot:
    plt.figure(figsize = (8, 6))

    sns.violinplot(
        data = plot_df,
        x = 'classification',
        y = feature,
        hue = 'classification',
        palette = 'Set2',
        legend = False
    )

    plt.title(f'Violin Plot of {feature.upper()}')
    plt.xlabel('Classification')
    plt.ylabel(feature.title())

    plt.tight_layout()

    # Save the plot
    plot_filename = f'violin_plot_{feature}.png'
    plot_path = os.path.join(output_dir, plot_filename)
    plt.savefig(plot_path, dpi = 300)

    plt.show()

print(f"Plots have been saved to: {output_dir}")
```

### 3.) Using the Ridge Plot. 
### Task 1: Provide the code for the ridge plot in the cell below

```python

import os
import seaborn as sns
import matplotlib.pyplot as plt

# Use the numerical data after imputation
plot_df = dfnum3.copy()

# Add the CKD classification labels
plot_df['classification'] = df['classification'].replace({
    0: 'Not CKD',
    1: 'CKD',
    'ckd': 'CKD',
    'notckd': 'Not CKD',
    'not ckd': 'Not CKD'
}).to_numpy()

# Directory where plots will be saved
output_dir = 'ridge_plots'
os.makedirs(output_dir, exist_ok = True)

# Select numerical features
features_to_plot = dfnum3.columns

# Set plot style
sns.set(font_scale = 1.2)
sns.set_style('white')

# Generate and save Ridge plots
for feature in features_to_plot:
    # Draw one density ridge for each CKD classification
    g = sns.FacetGrid(
        data = plot_df,
        row = 'classification',
        hue = 'classification',
        row_order = ['CKD', 'Not CKD'],
        palette = 'Set2',
        height = 1.7,
        aspect = 4,
        sharex = True,
        sharey = False
    )

    g.map_dataframe(
        sns.kdeplot,
        x = feature,
        fill = True,
        alpha = 0.75,
        linewidth = 1.2,
        warn_singular = False
    )

    # Add labels and baselines to the ridges
    for ax, label in zip(g.axes.flat, ['CKD', 'Not CKD']):
        ax.axhline(0, color = 'gray', linewidth = 0.8)
        ax.text(
            0.02, 0.25, label,
            transform = ax.transAxes,
            weight = 'bold'
        )
        ax.patch.set_alpha(0)

    g.set_titles('')
    g.set(yticks = [], ylabel = '')
    g.set_xlabels(feature.title())
    g.despine(left = True)
    g.fig.subplots_adjust(hspace = -0.25, top = 0.82)
    g.fig.suptitle(f'Ridge Plot of {feature.upper()}')

    # Save the plot
    plot_filename = f'ridge_plot_{feature}.png'
    plot_path = os.path.join(output_dir, plot_filename)
    g.fig.savefig(
        plot_path,
        dpi = 300,
        bbox_inches = 'tight'
    )

    plt.show()
    plt.close(g.fig)

print(f'Plots have been saved to: {output_dir}')
```
